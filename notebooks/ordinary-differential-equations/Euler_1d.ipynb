## Worked Example: Radioactive Decay

Consider a spontaneous radioactive decay of a particle with no external influence. Now, although the exact moment of the decay is random, the number of nuclei that will decay during a time interval $dt$ is $dN(t) \propto -N(t)\, dt$, where $N(t)$ is the number of nuclei present at time $t$. We can write it as a first-order differential equation: $dN(t)/dt = - \lambda N(t)$, where $\lambda$ is the decay rate. We know how to solve this equation analytically, but we will use Euler's method to find its numerical solution.

The numerical solution is given by: $N(i+1) = N(i) (1 - \lambda \Delta t), \quad i=0,1,...,N$.

Now, we will write a Python code to implement the above algorithm.

### Import Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Algorithm Implementation
Next, we define a function implementing Euler's method, which takes five arguments: 
- N0: initial number of nuclei
- t0: initial time
- dec_rate: decay rate
- dt: time step
- num_steps: number of steps

```python
# Define a function for Euler's algorithm
def euler(N0, t0, dec_rate, dt, num_steps):
    # Store time grid
    t = t0 + dt * np.arange(num_steps + 1)
    
    # Create an empty array and store N0 as the initial element
    y = np.empty(num_steps + 1, dtype=float)
    y[0] = N0

    # Run each iteration and store the values
    for i in range(num_steps):
        y[i+1] = y[i] * (1.0 - dec_rate * dt)
       
    return t, y  
```

### Initialization
With the function defined, we set up the initial conditions and parameters

```python
N0 = 1000   # Initial number of nuclei
dec_rate = 1.0   # Decay rate
ini_time = 0.0   # Initial time in seconds
fin_time = 10.0   # Final time in seconds
deltaT = 0.2   # Time interval (step size)

num_steps = int((fin_time - ini_time)/deltaT)
```

### Executing the Numerical and Analytical Solution

```python
time, euler_sol = euler(N0, ini_time, dec_rate, deltaT, num_steps)   # Call 'euler' function
analy_sol = N0 * np.exp(-dec_rate * (np.array(time) - ini_time)) # Calculate the analytical solution
```

### Visualize the result
We will now plot this and compare it with the analytical solution:

```python
# Plot numerical and analytical solution curves
plt.plot(time, euler_sol, "o--", label="Euler's Method, h=0.01 s")
plt.plot(time, analy_sol, "-", label="Analytical Solution")
plt.xlim(0, 5) # Plot only first 5 seconds 
plt.xlabel("Time (s)")
plt.ylabel("Number of Nuclei")
plt.title("Radioactive Decay using Euler's Method")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

We see that the numerical solution agrees well with the analytical solution within errors.
